### Conclusiones y observaciones:
1. **Inferencia:** El compilador de Scala es capaz de reconocer automáticamente qué tipo de dato es cada valor (por ejemplo, los números enteros por ejemplo asignaria a `Int` y los decimales a `Double`).
2. **tipos numéricos:** Para el `Float` es necesario añadir la letra `f` al final del número decimal (ej. `0.9f`), de lo contrario Scala lo interpretará como un `Double`.
3. **val vs var:** Los datos identificativos que sean estáticos como el nombre o los valores fijos iniciales se declaran como `val` (siendo inmutables), ya que no está previsto que cambien de forma normal durante la ejecución normal de estas variables base.

### Ejercicio 4 — Función para determinar si una mano se pasa de 21 

```scala
// Definición de la función pura 'bust'
def bust(puntuacion: Int): Boolean = {
  puntuacion > 21
}

// Pruebas 
val prueba18 = bust(18)
val prueba21 = bust(21)
val prueba22 = bust(22)
val prueba30 = bust(30)

println("Mano 18 se pasa (>21): " + prueba18)
println("Mano 21 se pasa (>21): " + prueba21)
println("Mano 22 se pasa (>21): " + prueba22)
println("Mano 30 se pasa (>21): " + prueba30)
```

### Conclusiónes:
- La función devuelve de manera directa el resultado de la comparación lógica (`puntuacion > 21`).
- Al tratarse de una función pura cumple estrictamente con el requisito de no modificar ninguna variable externa solo procesando los valores que se le introducen.

### Ejercicio 5 — Comparación de dos manos

```scala
// Definición de la función maxHand 
def maxHand(a: Int, b: Int): Int = {
  if (a >= b) a else b
}

// Pruebas obligatorias con los valores solicitados
val prueba1 = maxHand(17, 19)
val prueba2 = maxHand(20, 18)
val prueba3 = maxHand(21, 21)

// Mostramos los resultados por pantalla
println("Entre 17 y 19, el mayor es: " + prueba1)
println("Entre 20 y 18, el mayor es: " + prueba2)
println("Entre 21 y 21 (empate), el mayor es: " + prueba3)
```

### Conclusión y análisis de empates:
- **¿Qué sucede cuando ambas manos tienen el mismo valor?** Si la condición original hubiera sido estrictamente mayor (`if (a > b)`), al compararse con valores iguales (como `21 y 21`) el resultado daría falso y la función devolvería por defecto el segundo parámetro (`b`).
- **Decisión de mejora:** Para mejorar este problema he modificado la condición lógica usando mayor o igual (`a >= b`). De este modo, ante un empate, la función devuelve de forma predecible y controlada el primer valor comparado.

### Ejercicio 6 — Decidir el ganador de una partida

```scala
def ganador(handA:Int, handB:Int):Int = {
    val mayorA = bust(handA)
    val mayorB = bust(handB)
    if (mayorA && mayorB){
        0}//ambos superan 0
    else if(mayorA && !mayorB){
        handB}//si se pasa a gana b
    else if (!mayorA  && mayorB){
        handA}//si se pasa b gana a
    else{
        if (handA>= handB) handA else handB // si nadie supera 21 el que tenga mayor puntuacion gana 
    }
}



// Pruebas 
val res1 = ganador(26, 20)
val res2 = ganador(18, 22)
val res3 = ganador(24, 25)
val res4 = ganador(17, 19)
val res5 = ganador(21, 20)

println("Ganador entre 26 y 20: " + res1)
println("Ganador entre 18 y 22: " + res2)
println("Ganador entre 24 y 25: " + res3)
println("Ganador entre 17 y 19: " + res4)
println("Ganador entre 21 y 20: " + res5)
```

### Condiciones cumplidas en las pruebas:

1. **Prueba `ganador(26, 20)` — Resultado obtenido: `20`**
   - *Condición cumplida:* Se cumple el bloque `else if (mayorA && !mayorB)`. La mano A (26) supera 21 (`true`), pero la mano B (20) no se pasa (`false`). Como resultado, gana `handB` (`20`).

2. **Prueba `ganador(18, 22)` — Resultado obtenido: `18`**
   - *Condición cumplida:* Se cumple el bloque `else if (!mayorA && mayorB)`. La mano A (18) es válida, pero la mano B (22) supera 21. Como resultado, gana `handA` (`18`).

3. **Prueba `ganador(24, 25)` — Resultado obtenido: `0`**
   - *Condición cumplida:* Se cumple la primera condición `if (mayorA && mayorB)`. Ambas manos superan el límite de 21, por lo que el juego devuelve `0`.

4. **Prueba `ganador(17, 19)` — Resultado obtenido: `19`**
   - *Condición cumplida:* Se alcanza el bloque `else` final, ya que ninguna de las dos manos supera 21. Mediante la subcomparación interna `if (handA >= handB)`, se evalúa cuál de las dos puntuaciones válidas es mayor, resultando ganadora la mano de `19`.

5. **Prueba `ganador(21, 20)` — Resultado obtenido: `21`**
   - *Condición cumplida:* También se alcanza el bloque `else` final al ser ambos menores o iguales a 21. Al evaluar la expresión `21 >= 20`, la condición se cumple y devuelve la puntuación de `handA` (`21`).

## Ejercicio 7 — Arrays y mutabilidad

```scala
// Creamos el array de jugadores
val jugadores = Array("Alex", "Chen", "Marta")

// Mostramos el array original usando mkString para que no no devuelva una cadena no legible
println("Array original: " + jugadores.mkString(", "))

// Sustituimos "Alex" por "Sindhu"
jugadores(0) = "Sindhu"

// Mostramos el array con la modificacion
println("Array modificado: " + jugadores.mkString(", "))

// Error
//jugadores(1) = 500
```

### Documentación y respuestas teóricas:

1. **¿Por qué los elementos del array pueden cambiar aunque la variable se haya declarado con `val`?**
   - La palabra clave `val` asegura que la *referencia* a la variable `jugadores` no puede cambiar (es decir esta variable no puede apuntar a otro string distinto). pero el contenido interno sí esta permitidola alteracion los elementos que contiene.

2. **¿Por qué Scala no permite introducir el valor 500 en un `Array[String]`?**
   - Scala es un lenguaje estático. Como el array fue declarado para almacenar cadenas de texto`strings` no se permite almacenar datos numericos para evitar errores durante la ejecucion del programa

3. **Diferencia entre reasignar la variable y modificar un elemento:**
   - **Reasignar la variable** implicaría cambiar la estructura entera apuntando a un nuevo array (lo cual prohíbe el `val`).
   - **Modificar un elemento** consiste en alterar una posición ya existente dentro de la misma estructura de memoria que el array ya guardaba.

## Ejercicio 9 — Recorrer un Array con while

```scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0
while (i < manos.length) {
  val valorMano = manos(i)
  val resultadoBust = bust(valorMano)
  
  println("Posición " + i + " ---> Mano: " + valorMano + " | ¿Se pasa de 21?: " + resultadoBust)
  
  // Incrementamos el contador para avanzar a la siguiente posición hasta llegar el fin con el lenght
  i += 1
}
```

### respuestas teoricas

1. **¿Por qué el contador debe ser mutable (`var`)?**
   - Porque su valor necesita cambiar y actualizarse de forma constante en cada iteración del bucle (`i += 1`). Si se declarara como `val` (inmutable), no permitiría modificar su valor y el contador se quedaría bloqueado.

2. **¿Qué ocurriría si no incrementaras `i`?**
   - Se provocaría un **bucle infinito**. Como `i` valdría siempre `0`, la condición `i < manos.length` siempre sería verdadera, haciendo que el programa entrando en un bucle infinito repitiendo la posición 0 indefinidamente hasta agotar la memoria del equipo.

3. **¿Qué elementos del ejercicio corresponden al estilo imperativo?**
   - El uso de un  bucle tradicional `while` y y el uso de la variable mutable(`var i = 0`).Si se usara un estilo funciona usariamos métodos declarativos como `foreach` o `map`) el estilo imperativo detalla paso a paso y permite modificar variables 

## Ejercicio 10 — Listas e inmutabilidad

```scala
val jugadores = List("Alex", "Chen", "Marta")

println("Lista original: " + jugadores)

val jugadoresAñadidos= "Shindu"::jugadores

println("Nueva lista con 'Sindhu' añadida: " + jugadoresAñadidos)

println("Lista original:" + jugadores)

println("Longitud de la nueva lista: " + jugadoresAñadidos.length)

println("Lista invertida: " + jugadoresAñadidos.reverse)
```

### Respuestas

- **¿Por qué la operación con `::` produce una lista nueva en lugar de modificar la original?**
  - Para scala las listas son inmutables por lo que en vez de cambiar un objeto en memoria almacenado  el operador `::`crea una nueca lista que utiliza todo lo de la anterior y añade lo nuevo pero conserva la lista original

## Ejercicio 12 — Operadores relacionales y lógicos

```scala
val handA = 18
val handB = 21
val handC = 25

val expresion1 = handA > handB
val expresion2 = handB ==21 
val expresion3 = handC != 21
val expresion4 = (handA <= 21) && (handB <= 21)
val expresion5 = (handA > 21) || (handC > 21)
val expresion6 = !(handB > 21)

println("1. ¿handA > handB?: " + expresion1)
println("2. ¿handB == 21?: " + expresion2)
println("3. ¿handC != 21?: " + expresion3)
println("4. ¿handA y handB <= 21?: " + expresion4)
println("5. ¿Alguna de handA o handC > 21?: " + expresion5)
println("6. ¿No se ha pasado handB de 21?: " + expresion6)
```

### Tabla resumen de expresiones y resultados:

| Expresión | Resultado | Explicación |
| :--- | :---: | :--- |
| `handA > handB` | `false` | Evalúa si 18 es mayor que 21, lo cual es falso. |
| `handB == 21` | `true` | Comprueba la igualdad exacta entre 21 y 21. |
| `handC != 21` | `true` | Verifica que 25 sea diferente de 21. |
| `(handA <= 21) && (handB <= 21)` | `true` | Ambas condiciones se cumplen simultáneamente (`&&`), ya que 18 y 21 son menores o iguales a 21. |
| `(handA > 21) or (handC > 21)` | `true` | Al menos una se cumple (`or`): handA no supera 21, pero handC (25) sí. |
| `!(handB > 21)` | `true` | Como 21 no es mayor que 21 (`false`), el operador de negación `!` invierte el valor lógico a verdadero. |

## Ejercicio 13 — `foreach` y funciones como valores

```scala
val manos = Array(17, 24, 21, 26, 18)

println("version A")
var i =0 
while (i< manos.length){
    val mano = manos(i)
    println("Mano " +  mano + " se pasa? " + bust(mano))
    i+=1
}

println("Version B")
manos.foreach(mano=>{
    println("Mano " + mano  +" Se pasa? "+ bust(mano))
})
```

### Comparación entre versiones:

1. **¿Cuál necesita un contador?**
   - La **Versión A (`while`)** necesita obligatoriamente un contador numérico para saber en qué posición exacta del array se encuentra en cada momento.

2. **¿Cuál necesita una variable `var` para recorrer la colección?**
   - La **Versión A (`while`)** requiere declarar una variable mutable (`var i = 0`) para poder incrementar el índice de forma manual. La Versión B (`foreach`) no necesita ninguna variable externa .

3. **¿Cuál se aproxima más al estilo funcional?**
   - La **Versión B (`foreach`)** es completamente funcional y declarativa. En vez de tener que decirle como recorrer el array delegamos la tarea al metodo`for each`que procesa todos los elementos

## Ejercicio 14 — Efectos secundarios y estilo de programación


```scala
// --- VERSIÓN 1: Función con efecto secundario ---
var total = 0

def sumarAlTotal(valor: Int): Unit = {
  total = total + valor // Modifica una variable externa
}

println("--- Versión 1: Con efecto secundario ---")
println("Total inicial: " + total)

sumarAlTotal(5)
println("Después de sumar 5: " + total)

sumarAlTotal(10)
println("Después de sumar 10: " + total)

println()

// --- VERSIÓN 2: Función pura (sin efectos secundarios) ---
def sumar(a: Int, b: Int): Int = {
  a + b // Devuelve un resultado sin alterar nada externo
}

println("--- Versión 2: Función pura ---")
val resultado1 = sumar(5, 10)
val resultado2 = sumar(20, 30)

println("Resultado de sumar(5, 10): " + resultado1)
println("Resultado de sumar(20, 30): " + resultado2)
```

### Explicación:

1. **¿Por qué la función modifica una variable situada fuera de su ámbito local?**
   - Porque `sumarAlTotal` accede a una variable externa (`var total`) modificando su valor directamente en lugar en vez de  trabajar  con los parámetros que recibe.

2. **Identificación del efecto secundario:**
   - El **efecto secundario** es precisamente la alteración del programa fuera de los límites de la función (modifica el valor de `total`). 

### Tabla comparativa de funciones:

| Característica | `sumarAlTotal` | `sumar` |
| :--- | :---: | :---: |
| **Modifica datos externos** | Sí | No |
| **Devuelve un resultado calculado** | No (devuelve `Unit`) | Sí (`Int`) |
| **Utiliza efecto secundario** | Sí | No |
| **Estilo predominante** | Imperativo  | Funcional |

## Ejercicio 15 — Programa integrado: Torneo de Twenty-One

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manosRonda1 = Array(18, 24, 21, 20)
val manosRonda2 = Array(22, 19, 20, 21)
//funcion que comprueba la maxima puntuacion y si es valida 
def maxPuntuacionValida(manos: Array[Int]): Int = {
    var maxValido = 0
    manos.foreach(m => {
        if (!bust(m) && m > maxValido) {
            maxValido = m
        }
    })
    maxValido 
}


println("Ronda 1")
var i = 0
//mediante un bucle while recorremos toda la lista llamando a la funcion en cada valor
while (i < jugadores.length) {
    val jugador = jugadores(i)
    val mano = manosRonda1(i)
    val estado = if (bust(mano)) "Se pasa" else "Válida"
    
    println("Jugador " +  jugador + " Mano " + mano +" Estado " + estado)
    i += 1
    
}
val mejorR1 = maxPuntuacionValida(manosRonda1)
println( "Mejor puntuación válida Ronda 1:" +mejorR1 )

println("Ronda 2")
//en vez de un while utilizamos zip para vincular jugadores con su puntuacion haciendo algo similar a un dicionario de python 
//con un for each lo recoremos entero caso por caso calculando si se paso o no con bust  
jugadores.zip(manosRonda2).foreach { case (jugador, mano) =>
    val estado = if (bust(mano)) "Se pasa" else "Válida"
    println(s"Jugador: $jugador  Mano: $mano  Estado: $estado")
}
val mejorR2 = maxPuntuacionValida(manosRonda2)
//s equivalente a los f-strings de python 
println(s"Mejor puntuación válida Ronda 2: $mejorR2")
```
